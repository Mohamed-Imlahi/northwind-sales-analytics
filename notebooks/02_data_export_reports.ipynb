```python


import pandas as pd
import sqlite3
conexion= sqlite3.connect('northwind.db')



# 1. Catálogo unificado de productos (Top 10 más caros)
query_catalogo = """
SELECT p.ProductID, p.ProductName, s.CompanyName, c.CategoryName, p.UnitPrice
FROM Products p
INNER JOIN Suppliers s ON s.SupplierID = p.SupplierID
INNER JOIN Categories c ON c.CategoryID = p.CategoryID
WHERE p.Discontinued = 0
ORDER BY p.UnitPrice DESC
LIMIT 10;
"""
df_catalogo_productos = pd.read_sql(query_catalogo, conexion)
df_catalogo_productos.to_csv('catalogo_productos_top10.csv', index=False, encoding='utf-8-sig')

# 2. Auditoría de clientes inactivos
query_inactivos = """
SELECT c.CustomerID, c.CompanyName, c.ContactName, c.Country 
FROM Customers c
LEFT JOIN Orders o ON o.CustomerID = c.CustomerID
WHERE o.CustomerID IS NULL;
"""
df_clientes_inactivos = pd.read_sql(query_inactivos, conexion)
df_clientes_inactivos.to_csv('clientes_inactivos.csv', index=False, encoding='utf-8-sig')


# 3. Rendimiento comercial por empleado (Top 10)
query_rendimiento = """
SELECT e.EmployeeID, e.FirstName || ' ' || e.LastName AS Nombre_completo,
       COUNT(DISTINCT o.OrderID) AS total_pedidos, 
       ROUND(SUM(od.Quantity * od.UnitPrice * (1 - od.Discount)), 2) AS Facturacion_Total
FROM Employees e
LEFT JOIN Orders o ON o.EmployeeID = e.EmployeeID
LEFT JOIN "Order Details" od ON od.OrderID = o.OrderID
GROUP BY e.EmployeeID, e.FirstName, e.LastName
ORDER BY Facturacion_Total DESC
LIMIT 10;
"""
df_rendimiento_empleados = pd.read_sql(query_rendimiento, conexion)
df_rendimiento_empleados.to_csv('rendimiento_empleados.csv', index=False, encoding='utf-8-sig')

```
